```sql
%python
'''
OPTIMIZE

OPTIMIZE is a Delta Lake optimization technique used to address the small-file problem. It compacts multiple small files into fewer larger files, reducing metadata overhead, file-open costs, and improving query performance.

It is especially useful after frequent inserts, streaming ingestion, MERGE, UPDATE, or DELETE operations.

Key point: OPTIMIZE improves file compaction but does not organize data for filtering or improve data skipping on its own. OPTIMIZE does not change the logical order of data.

Benefits

✅ Faster reads

✅ Lower metadata overhead

✅ Better Spark task utilization

✅ Lower cloud storage API calls

What OPTIMIZE does NOT do

❌ Doesn't partition data

❌ Doesn't sort data

❌ Doesn't improve data skipping by itself
---------------------------------------------------------
e.g

OPTIMIZE sales;

Before:
1000 small files

After:
20 large files
'''
```

```sql
%python
'''
ZORDER

ZORDER is a data layout optimization technique, that improves the data skipping. It hysically reorganizes data so similar values are stored closer together, so that Databricks can read fewer files during query execution.

Without ZORDER, data can be scattered across many files.
It is most beneficial for columns that are frequently used in WHERE clauses, JOINs, and GROUP BY operations.
ZORDER is executed as part of an OPTIMIZE operation.

Key points:

- Improves query performance through data skipping.
- Reduces the amount of data scanned.
- New incoming data is not automatically re-Z-ordered.
- Therefore, it typically needs to be rerun periodically.

-----------------------------------------------------------------------------------------------------------------
e.g. 

OPTIMIZE sales ZORDER BY (customer_id);
'''
```

```sql
%python
'''
LIQUID Clustering

LIQUID Clustering is Databricks' modern data organization and optimization technique that improves data skipping and also reduces manual maintenance. 

Liquid clustering stores the clustering keys as table metadata. Databricks then uses those clustering keys during maintenance operations to maintain an efficient data layout.

Key points:

Advantages

✅ No partition explosion
✅ Easier maintenance, no manual maintenance
✅ Better than over-partitioning
✅ Works well with growing datasets
✅ data skipping
✅ Recommended by Databricks for new tables

--------------------------------------------------------------------------------------------------------------------

CREATE TABLE sales (
    id BIGINT,
    customer_id BIGINT,
    country STRING,
    amount DOUBLE
)
CLUSTER BY customer_id;


ALTER TABLE sales
CLUSTER BY (customer_id, country);
'''
```

```sql
%python
'''
AUTO Liquid Clustering

Databricks automatically selects clustering keys and evolves it over a time based on query patterns.


e.g. 

CREATE TABLE sales (
    id BIGINT,
    customer_id BIGINT,
    country STRING,
    amount DOUBLE
)
CLUSTER BY AUTO;


ALTER TABLE sales
CLUSTER BY AUTO;



'''
```
